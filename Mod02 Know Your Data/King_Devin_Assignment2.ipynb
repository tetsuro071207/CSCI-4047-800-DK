```python
# Code here
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("./Income Data.csv")
df.head()
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
# Code here

def func(column_name): 
    min = np.min(df[column_name])
    q1 = np.percentile(df[column_name], 25)
    med = np.median(df[column_name])
    q3 = np.percentile(df[column_name], 75)
    max = np.max(df[column_name])

    print(f'------------ Name:{column_name} -------------')
    print(f"Min: {min}")
    print(f"Q1: {q1}")
    print(f"Median: {med}")
    print(f"Q3: {q3}")
    print(f"Max: {max}")

func("income")
func("prestige")
func("education")

```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
# Code here
plt.hist(df['prestige'], bins=12, color='orange', alpha=0.5)
plt.show()
plt.hist(df['education'], bins=12, color='blue', alpha=0.5)
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
# Code here
plt.scatter(x=df['income'], y = df['education'], alpha=0.3)
plt.title("Incomes to Education")
plt.xlabel('income')
plt.ylabel("education")
plt.show()
plt.scatter(x=df['education'], y = df['prestige'], alpha=0.3)
plt.title("Education to Prestige")
plt.xlabel('education')
plt.ylabel("prestige")
plt.show()
```
